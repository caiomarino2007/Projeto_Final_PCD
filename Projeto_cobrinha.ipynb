#### *O projeto apresentado a seguir consiste em uma tentativa de réplica do clássico jogo da cobrinha, o "Snake", que pode ser acessado a qualquer momento pelo google.<br>Para a produção do código, foi utilizada a biblioteca "pygame", a qual é uma biblioteca de código aberto para a linguagem Python, criada para o desenvolvimento de jogos 2D.*

#### O jogo funciona a partir de um laço de repetição while, e é nele que, enquanto o jogo roda, a tela é atualizada de forma contínua, fazendo com que os eventos como movimentação da cobrinha, comer as maçãs e as colisões, sejam verificados e comandos do usuário sejam processados. A cobrinha é feita por uma lista de listas (que representam os segmentos do corpo - cabeça, corpo, cauda) que têm suas posições atualizadas quando uma tecla direcional é apertada. O jogo termina quando a cobrinha colidir com as bordas do campo ou com o próprio corpo.

##### A célula abaixo importa as bibliotecas que serão utilizadas ao longo do projeto e incializa o módulo pygame.<br>**OBS.: o comando que está em comentário é o comando que instala a biblioteca pygame no kernel do Python que está sendo utilizado no momento (execute-o caso você não tenha a biblioteca instalada no seu kernel)**

```python
# %pip install pygame
import pygame
import random

# Inciando o pygame
pygame.init()
```

###### O módulo random importado acima, é uma biblioteca nativa do próprio Python que pode gerar números pseudoaleatórios, ou seja, não são totalmente aleatórios, eles são gerados por algoritmos matemáticos. Portanto, ela será utilizada para "aleatorizar" a posição em que a maçã irá spawnar

##### A célula abaixo contém as variáveis constantes que serão utilizadas ao longo do projeto, como o tamanho da janela, posição e tamanho do campo (grid) de jogo, e o tamanho de um bloco no campo: 

```python
# Tamanho da janela
larg = 800
alt = 600

# Posição e tamanho do campo
campo_x = 100
campo_y = 50
larg_campo = 600
alt_campo = 500

# Tamanho do bloco
tam_bloco = 20
```

##### A célula abaixo é responsável por criar a janela que será o jogo e definir o nome para a mesma:

```python
# Criando a janela
janela = pygame.display.set_mode((larg, alt))

# Definindo nome para o jogo
pygame.display.set_caption("Jogo da Cobrinha")
```

##### A célula abaixo é responsável por criar o que representará o estado atual do jogo. É aqui em que é criado a variável que controla o loop do jogo, a que controla se o usuário perdeu ou não, a lista que representa a cobrinha, as variáveis que representam a velocidade da cobrinha, a sua pontuação, e a variável responsável por controlar quando a cobrinha come a maçã:

```python
# Variável de controle do loop principal
rodando = True

# Variável de controle se o usuário perdeu ou não
perdeu = False

# Cobrinha como lista de listas
cobrinha = [
    [140, 110],
    [120, 110],
    [100, 110]
]

# Variáveis de velocidade
vx = 0
vy = 0

# Contador de pontuação
pontuacao = 0

# Variável de controle ao comer a maçã
cobra_maca = False
```

#### Na célula a seguir, estão definidas as variáveis que serão utilizadas na diferenciação e implementação de dois modos de jogo: o modo padrão de jogo e o modo daltonismo. Dessa forma, o projeto se torna mais inclusivo, ao permitir que o público daltônico consiga distinguir as cores do jogo.

```python
# Variável que será usada para ligar/desligar modo daltonismo (por padrão, tem o modo daltonismo desligado)
modo_daltonismo = False

# Cores no modo não daltônico
cor_fundo = (152, 251, 152) # Cor verde-pálido
cor_cobrinha = (72, 110, 67) # Cor verde-escuro
cor_maca = (255, 8, 0) # Cor vermelho-maçã
cor_grid = (170, 220, 170) # Cor verde-claro
cor_grid_alt = (160, 210, 160) # Cor verde levemente escuro
cor_pontuacao = (139, 0, 0) # Cor vermelho escuro

# Cores no modo daltonismo
cor_fundo_dalt = (220, 220, 220) # Cor cinza-claro
cor_cobrinha_dalt = (30, 80, 200) # Cor azul royal
cor_maca_dalt = (255, 140, 0) # Cor laranja
cor_grid_dalt = (200, 200, 200) # Cor cinza-claro
cor_grid_alt_dalt = (180, 180, 180) # Cor cinza médio
cor_pontuacao_dalt = (20, 20, 20) # Cor cinza muito escuro
```

##### A célula abaixo é responsável por criar a função que gera uma posição aleatória para a maçã:

```python
def posicao_maca():
    """Por meio do módulo random, cria posições aleatórias para a maçã, de forma que
    ao ser chamada quando a cobrinha come a maçã, gera sempre uma posição aleatória
    para a mesma."""

    # Posição x da maçã
    x = campo_x + random.randint(1, (larg_campo//tam_bloco) - 2) * tam_bloco

    # Posição y da maçã
    y = campo_y + random.randint(1, (alt_campo//tam_bloco) - 2) * tam_bloco

    return x, y
```

##### A célula abaixo é responsável por criar a fonte que será utilizada para exibir a pontuação:

```python
# Criando fonte
fonte = pygame.font.Font(None, 30)
fonte_menor = pygame.font.Font(None, 20)
```

##### A função presente no código abaixo é responsável por reiniciar todas as variáveis de estado do jogo para seus valores originais. Ela será útil quando o usuário quiser reiniciar o jogo quando ele perder.

```python
def reinicia_jogo():
    """Retorna todas as variáveis para seus valores originais, de forma
    que volte para como deveria ser ao iniciar o jogo."""
    
    cobrinha = [
    [140, 110],
    [120, 110],
    [100, 110]
    ]

    cobra_maca = False
    maca_x, maca_y = posicao_maca()

    vx = 0
    vy = 0

    pontuacao = 0

    return cobrinha, cobra_maca, maca_x, maca_y, vx, vy, pontuacao
```

##### *OBS.: O pygame tem uma limitação que impede que o código de funcionamento do jogo seja programado em diferentes células do Jupyter, visto que o laço de repetição while que faz o jogo rodar é contínuo para atualizar a janela e processar os eventos. Portanto, a maior parte do código tem que ser executado de forma sequencial, por isso fica em uma única célula.*<br><br>**Na célula a seguir, o código presente na mesma contém o loop principal do jogo, ou seja, todos os eventos que serão atualizados e todos os comandos do usuário que serão processados estão contidos nesse loop. Portanto, essa célula é a principal de todo o projeto e é aqui que o jogo realmente acontece.**

```python
# Criando variáveis que representarão a posição da maça
maca_x, maca_y = posicao_maca()

# Criando variável para limitar o FPS (frames per second)
fps = pygame.time.Clock()

while rodando:

    # Loop de eventos
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT: # Ordena que o jogo feche quando o usuário apertar no X da janela.
            rodando = False
        if evento.type == pygame.KEYDOWN: # Condicional para caso alguma tecla seja pressionada
            if perdeu == False:
                if evento.key == pygame.K_RIGHT and vx != -20: # Mover para direita ao pressionar seta direita
                    vx = 20
                    vy = 0
                        
                elif evento.key == pygame.K_LEFT and vx != 20: # Mover para esquerda ao pressionar seta esquerda
                    vx = -20
                    vy = 0
                        
                elif evento.key == pygame.K_UP and vy != 20: # Mover para cima ao pressionar seta cima
                    vy = -20
                    vx = 0
                        
                elif evento.key == pygame.K_DOWN and vy != -20: # Mover para baixo ao pressionar seta baixo
                    vy = 20
                    vx = 0
                
                # Alternando entre modos
                elif evento.key == pygame.K_d: # Ativa modo daltonismo ao apertar tecla D
                    modo_daltonismo = True
                elif evento.key == pygame.K_c: # Desativa modo daltonismo ao apertar tecla C
                    modo_daltonismo = False

            if perdeu == True:
                if evento.key == pygame.K_ESCAPE: # Fecha o jogo ao apertar ESC
                    rodando = False
                if evento.key == pygame.K_r: # Reinicia o jogo ao apertar R
                    cobrinha, cobra_maca, maca_x, maca_y, vx, vy, pontuacao = reinicia_jogo()
                    perdeu = False

    if perdeu == False:
         
        # Aqui é o evento principal do jogo, que é quando a cobra colide/come a maçã
        if cobrinha[0][0] == maca_x and cobrinha[0][1] == maca_y:
            cobra_maca = True
            maca_x, maca_y = posicao_maca()
            pontuacao += 1

        # Alterando a posição da cobrinha a depender da tecla pressionada
        cabeca_x = cobrinha[0][0]
        cabeca_y = cobrinha[0][1]
        cresce_cabeca = [cabeca_x + vx, cabeca_y + vy]
        if vx != 0 or vy != 0:
            cobrinha.insert(0, cresce_cabeca)
            if cobra_maca == False:
                cobrinha.pop()
            else:
                cobra_maca = False
        
        cabeca_x = cobrinha[0][0]
        cabeca_y = cobrinha[0][1]

        # Fazendo colisão com a parede do campo fechar o jogo
        if cabeca_x <= campo_x:
            perdeu = True
        if cabeca_x >= campo_x + larg_campo - 20:
            perdeu = True
        if cabeca_y <= campo_y:
            perdeu = True
        if cabeca_y >= campo_y + alt_campo - 20:
            perdeu = True
        
        # Fazendo o jogo fechar quando a cobra bater nela mesma:
        if cobrinha[0] in cobrinha[1:]:
            perdeu = True

        # Texto de instrução para alternar modos
        texto_dalt = fonte_menor.render("Aperte 'D' para ativar o modo daltonismo!", True, (0, 0, 0))
        texto_n_dalt = fonte_menor.render("Aperte 'C' para desativar", True, (0, 0, 0))

        # Caso esteja no modo sem daltonismo
        if modo_daltonismo == False:
            # Renderizando o texto da pontuação do jogo
            texto_pontuacao = fonte.render(f'Pontuação: {pontuacao}', True, cor_pontuacao)

            # Pinta toda a tela com a cor verde-pálido (pale green).
            janela.fill(cor_fundo)

            # Desenhando o campo (grid) de jogo
            for linha in range(alt_campo//tam_bloco):
                for coluna in range(larg_campo//tam_bloco):
                    pos_x = campo_x + coluna * tam_bloco
                    pos_y = campo_y + linha * tam_bloco

                    # Fazendo quadrados de cores alternadas no campo (parecendo um tabuleiro de xadrez)
                    if (linha + coluna) % 2 == 0:
                        pygame.draw.rect(janela, cor_grid, (pos_x, pos_y, 20, 20))
                    else:
                        pygame.draw.rect(janela, cor_grid_alt, (pos_x, pos_y, 20, 20))

            # Desenhando a maçã
            pygame.draw.circle(janela, cor_maca, (maca_x + 10, maca_y + 10), 10)

            # Desenhando a cobrinha
            for corpo in cobrinha:
                corpo_x = corpo[0]
                corpo_y = corpo[1]
                pygame.draw.rect(janela, cor_cobrinha, (corpo_x, corpo_y, 20, 20))

        # Caso esteja no modo daltonismo
        if modo_daltonismo == True:
            
            # Renderização pontuação
            texto_pontuacao = fonte.render(f'Pontuação: {pontuacao}', True, cor_pontuacao_dalt)
            
            # Pintando fundo
            janela.fill(cor_fundo_dalt)
            
            # Desenhando campo
            for linha in range(alt_campo//tam_bloco):
                for coluna in range(larg_campo//tam_bloco):
                    pos_x = campo_x + coluna * tam_bloco
                    pos_y = campo_y + linha * tam_bloco

                    # Fazendo cores alternadas no campo
                    if (linha + coluna) % 2 == 0:
                        pygame.draw.rect(janela, cor_grid_dalt, (pos_x, pos_y, 20, 20))
                    else:
                        pygame.draw.rect(janela, cor_grid_alt_dalt, (pos_x, pos_y, 20, 20))

            # Desenhando maçã
            pygame.draw.circle(janela, cor_maca_dalt, (maca_x + 10, maca_y + 10), 10)

            # Desenhando cobra
            for corpo in cobrinha:
                corpo_x = corpo[0]
                corpo_y = corpo[1]
                pygame.draw.rect(janela, cor_cobrinha_dalt, (corpo_x, corpo_y, 20, 20))
        
        # Desenhando o contorno/borda do campo 
        pygame.draw.rect(janela, (0, 0, 0), (campo_x, campo_y, larg_campo, alt_campo), 5)

        # Desenhando pontuação do jogo
        janela.blit(texto_pontuacao, (60, 20))

        # Desenhando instrução de alternar modos
        janela.blit(texto_dalt, (520, 15))
        janela.blit(texto_n_dalt, (520, 30))

    # Quando o usuário perder
    else:
        if modo_daltonismo == False:
            texto_perdeu = fonte.render("Você perdeu!", True, (255, 0, 0)) # Texto perdeu vermelho
        else:
            texto_perdeu = fonte.render("Você perdeu!", True, (255, 215, 0)) # Texto perdeu dourado
        texto_pontuacao_p = fonte.render(f'Pontuação: {pontuacao}', True, (200, 200, 200))
        texto_reinciar_p = fonte.render("Reiniciar: R", True, (255, 255, 255))
        texto_fechar_p = fonte.render("Fechar: ESC", True, (255, 255, 255))

        janela.fill((0, 0, 0)) # Pintando a tela toda de preto
        janela.blit(texto_perdeu, (320, 200)) # Desenhando o texto perdeu
        janela.blit(texto_pontuacao_p, (320, 250)) # Desenhando a pontuação que o usuário fez
        janela.blit(texto_reinciar_p, (320, 320)) # Desenhando instrução para reiniciar
        janela.blit(texto_fechar_p, (320, 360)) # Desenhando instrução para fechar

    # Atualizando a tela do jogo
    pygame.display.update()

    # Limitando a janela a 15 FPS
    fps.tick(15)
    
    

pygame.quit()
```

#### **Abaixo, segue todas as funções da biblioteca pygame e random utilizadas no projeto devidamente explicadas.**

#### Pygame: <br>
##### **pygame.init()**: Inicializa o módulo pygame. Sem executar essa função, várias outras funções da biblioteca não funcionam; <br> **pygame.display.set_mode()**: Cria a janela em que o jogo será executado. <br> **pygame.display.set_caption()**: Define o nome da janela e recebe como argumento uma string. <br> **pygame.display.update()**: Responsável por atualizar a tela, fazendo com que tudo o que foi desenhado apareça de fato na janela criada. <br> **pygame.event.get()**: Obtém todos os eventos ocorridos desde a última atualização. Ela captura e limpa a fila de eventos. <br> **pygame.draw.rect()**: Desenha um retângulo e recebe pelo menos três argumentos, sendo eles a superfície em que o retângulo será desenhado, a cor do retângulo em uma tupla de valores RGB, e a dimensão e posição do retângulo, respectivamente. Um último argumento pode ser adicionado para desenhar uma borda, e esse argumento representa a espessura dessa borda. <br> **pygame.draw.circle()**: Desenha um círculo e recebe pelo menos quatro argumentos, sendo eles a superfície em que o círculo será desenhado, a cor do círculo em uma tupla de valores RGB, a posição do centro do círculo, e o raio. <br> **.fill()**: Preenche toda a superfície da janela com uma cor específica (em tupla RGB) que é o argumento da função. <br> **pygame.font.Font()**: Cria uma fonte e recebe como argumento a fonte e o tamanho da mesma, respectivamente. <br> **fonte.render()**: Renderiza a fonte, transformando o texto em imagem. Recebe como argumento o texto (em string), o antialiasing (antiserrilhamento; deve ser True ou False), e a cor (tupla RGB), respectivamente. <br> **.blit()**: Desenha o texto na tela. Recebe como argumento o texto renderizado e a posição que ficará o mesmo, respectivamente. <br> **pygame.time.Clock()**: Cria um relógio que controla a velocidade do jogo (é usado para controlar o FPS). <br> **Clock.tick()**: Limita o FPS (quadros por segundo) ao valor que é inserido como argumento. <br> **pygame.quit()**: Finaliza o pygame, ou seja, fecha todos os módulos inicializados pela pygame.init(). <br>
#### Random: <br>
##### **random.randint()**: Retorna um número inteiro aleatório entre o número inserido como primeiro argumento e o número inserido como segundo argumento.
