##### A célula abaixo importa as bibliotecas que serão utilizadas ao longo do projeto.<br>**OBS.: o comando que está em comentário é o comando que instala a biblioteca pygame no kernel do Python que está sendo utilizado no momento**

```python
# %pip install pygame
import pygame
import random
import sys
```

##### **Na célula a seguir, o código presente na mesma executa o pygame e faz o jogo rodar.**

```python
# O comando abaixo inicializa o pygame, ou seja, inicializa o jogo.
pygame.init()


# Definindo a altura e largura da janela que executará o jogo
larg = 800
alt = 600

# Criando a janela e definindo nome, respectivamente.
janela = pygame.display.set_mode((larg, alt))
pygame.display.set_caption("Jogo da Cobrinha")

"""Criando variável para controlar o loop do jogo.
Esse loop é onde tudo que acontece no jogo será programado, ele mantém a tela 'viva',
atualizando o jogo, a tela e os comandos."""
rodando = True

while rodando:
    # O comando pygame.event.get() é responsável por capturar e limpar a fila de eventos
    # do pygame. Ele lista todas as interações que aconteceram desde a última vez que a função foi chamada.
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT: # Ordena que o jogo feche quando o usuário apertar no X da janela.
            rodando = False

    # Pinta toda a tela com a cor verde-pálido (pale green). OBS.:Recebe como argumento uma tupla de valores RGB.
    janela.fill((152, 251, 152))
    # O comando 'pygame.display.update()' é responsável por atualizar a janela do jogo na tela.
    pygame.display.update()

pygame.quit()
```
