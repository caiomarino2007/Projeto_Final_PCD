#### *O projeto apresentado a seguir consiste em um réplica do clássico jogo da cobrinha, o "Snake", que pode ser acessado a qualquer momento pelo google.<br>Para a produção do código, foi utilizada a biblioteca "pygame", a qual é uma biblioteca de código aberto para a linguagem Python, criada para o desenvolvimento de jogos 2D.*

##### A célula abaixo importa as bibliotecas que serão utilizadas ao longo do projeto.<br>**OBS.: o comando que está em comentário é o comando que instala a biblioteca pygame no kernel do Python que está sendo utilizado no momento**

```python
# %pip install pygame
import pygame
import random
```

##### **Na célula a seguir, o código presente na mesma executa o pygame e faz o jogo rodar.**

```python
# O comando abaixo inicializa o pygame, ou seja, inicializa o jogo.
pygame.init()


# Definindo a altura e largura da janela que executará o jogo
larg = 800
alt = 600

# Criando a janela e definindo nome, respectivamente.
janela = pygame.display.set_mode((larg, alt))
pygame.display.set_caption("Jogo da Cobrinha")

# Criando variáveis que representarão posição e tamanho, respectivamente, do "campo"
# onde a cobrinha se movimentará
campo_x = 100
campo_y = 50
larg_campo = 600
alt_campo = 500

"""Criando variável para controlar o loop do jogo.
Esse loop é onde tudo que acontece no jogo será programado, ele mantém a tela 'viva',
atualizando o jogo, a tela e os comandos."""
rodando = True

# Criando o corpo da cobrinha (lista de listas (partes do corpo))
cobrinha = [
    [140, 110],
    [120, 110],
    [100, 110]
]

# Criando variáveis que representarão o movimento da cobrinha
vx = 0
vy = 0

# Criando variáveis que representarão a posição da maça
maca_x = campo_x + random.randint(1, (larg_campo//20) - 2) * 20
maca_y = campo_y + random.randint(1, (alt_campo//20) - 2) * 20

# Criando variável para limitar o FPS (frames per second)
fps = pygame.time.Clock()

while rodando:
    # O comando pygame.event.get() é responsável por capturar e limpar a fila de eventos
    # do pygame. Ele lista todas as interações que aconteceram desde a última vez que a função foi chamada.
    for evento in pygame.event.get():
        if evento.type == pygame.QUIT: # Ordena que o jogo feche quando o usuário apertar no X da janela.
            rodando = False
        if evento.type == pygame.KEYDOWN: # Condicional para caso alguma tecla seja pressionada
            if evento.key == pygame.K_RIGHT: # Mover para direita ao pressionar seta direita
                vx = 20
                vy = 0
            elif evento.key == pygame.K_LEFT: # Mover para esquerda ao pressionar seta esquerda
                vx = -20
                vy = 0
            elif evento.key == pygame.K_UP: # Mover para cima ao pressionar seta cima
                vy = -20
                vx = 0
            elif evento.key == pygame.K_DOWN: # Mover para baixo ao pressionar seta baixo
                vy = 20
                vx = 0

    # Alterando a posição da cobrinha a depender da tecla pressionada
    cabeca_x = cobrinha[0][0]
    cabeca_y = cobrinha[0][1]
    cresce_cabeca = [cabeca_x + vx, cabeca_y + vy]
    if vx != 0 or vy != 0:
        cobrinha.insert(0, cresce_cabeca)
        cobrinha.pop()
        
    # Aqui é o evento principal do jogo, que é quando a cobra colide/come a maçã
    if cobrinha[0][0] == maca_x and cobrinha[0][1] == maca_y:
        maca_x = campo_x + random.randint(1, (larg_campo//20) - 2) * 20
        maca_y = campo_y + random.randint(1, (alt_campo//20) - 2) * 20

    # Pinta toda a tela com a cor verde-pálido (pale green). OBS.:Recebe como argumento uma tupla de valores RGB.
    janela.fill((152, 251, 152))

    # Desenhando o campo (grid) de jogo
    for linha in range(alt_campo//20):
        for coluna in range(larg_campo//20):
            pos_x = campo_x + coluna * 20
            pos_y = campo_y + linha * 20

            # Fazendo quadrados de cores alternadas no campo (parecendo um tabuleiro de xadrez)
            """A função abaixo é responsável por desenhar um retângulo na tela e recebe
             três argumentos: o primeiro é a janela em que o retângulo será desenhado;
            o segundo é a cor (tupla RGB) que será o retângulo; e o último é a dimensão e posição do retângulo"""
            if (linha + coluna) % 2 == 0:
                pygame.draw.rect(janela, (170, 220, 170), (pos_x, pos_y, 20, 20))
            else:
                pygame.draw.rect(janela, (160, 210, 160), (pos_x, pos_y, 20, 20))

    # Desenhando o contorno/borda do campo
    pygame.draw.rect(janela, (0, 0, 0), (campo_x, campo_y, larg_campo, alt_campo), 5)

    # Desenhando o esqueleto da maçã para teste
    pygame.draw.rect(janela, (255, 8, 0), (maca_x, maca_y, 20, 20))

    # Desenhando a cobrinha
    for corpo in cobrinha:
        corpo_x = corpo[0]
        corpo_y = corpo[1]
        pygame.draw.rect(janela, (72, 110, 67), (corpo_x, corpo_y, 20, 20))

    # O comando 'pygame.display.update()' é responsável por atualizar a janela do jogo na tela.
    pygame.display.update()

    # Limitando a janela a 15 FPS
    fps.tick(15)

pygame.quit()
```
